```python
# Load from .env and pass credentials explicitly to subset()
import os
from dotenv import load_dotenv

load_dotenv()  # loads .env in current working directory

USERNAME = os.getenv("COPERNICUS_MARINE_USERNAME")
PASSWORD = os.getenv("COPERNICUS_MARINE_PASSWORD")
```

```python
import shutil
import copernicusmarine

OUTPUT_DIR = "raw_data"
if os.path.isdir(OUTPUT_DIR):
    shutil.rmtree(OUTPUT_DIR)
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"Reset {OUTPUT_DIR}/ — downloading fresh files")

MIN_LON = 87
MAX_LON = 91
MIN_LAT = 20
MAX_LAT = 23
START_DATETIME = "2020-01-01T00:00:00"
END_DATETIME = "2026-12-31T00:00:00"
MIN_DEPTH = 0.5
MAX_DEPTH = 0.5
```

```python
output_path = f"{OUTPUT_DIR}/thetao.nc"

copernicusmarine.subset(
    dataset_id="cmems_mod_glo_phy-thetao_anfc_0.083deg_P1D-m",
    variables=["thetao"],
    minimum_longitude=MIN_LON,
    maximum_longitude=MAX_LON,
    minimum_latitude=MIN_LAT,
    maximum_latitude=MAX_LAT,
    start_datetime=START_DATETIME,
    end_datetime=END_DATETIME,
    minimum_depth=MIN_DEPTH,
    maximum_depth=MAX_DEPTH,
    output_filename=output_path,
    username=USERNAME,
    password=PASSWORD,
)

print("Download completed:", output_path)
```
